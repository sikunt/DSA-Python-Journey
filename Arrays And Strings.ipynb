## 1. Reverse String
#### Write a function that reverses a string. The input string is given as an array of characters s.
#### You must do this by modifying the input array in-place with O(1) extra memory.

### Example 1:
#### Input: s = ["h","e","l","l","o"]
#### Output: ["o","l","l","e","h"]

### Example 2:
#### Input: s = ["H","a","n","n","a","h"]
#### Output: ["h","a","n","n","a","H"]

### Constraints:

* 1 <= s.length <= 105
* s[i] is a [printable ascii character](https://en.wikipedia.org/wiki/ASCII#Printable_characters).



Python (Two Pointers, Iteration, O(1) Space)
* Time complexity : O(N) to swap N/2 element
* Space complexity : O(1), it's a constant space solution

```python
# Do not return anything, modify s in-place instead.
# Hint: The entire logic for reversing a string is based on using the opposite directional two-pointer approach!
        
class Solution: # Common practice to group related methods within class
    def reverseString(self, s:list[str]) -> None: # methods (function) within class takes 'self'(instance of class) and 's' parameters, return type hint indicates method returns no
        left = 0 # initialize first pointer to start of list
        right = len(s) - 1 # initialize second pointer to end of list

        while left < right: # start while loop which runs until the end of list is reached
            s[left], s[right] = s[right], s[left] # Swap elements of the two pointers
            left += 1 # Move left pointer right
            right -= 1 # Move right pointer left
        
    # Method ends after loop reverses list in place
```

Python (Recursion, In-Place O(N) Space)
* in-place algorithms transform inputs using no auxiliary data structure and can have recursion stacks that are NOT constant
    
    

* Time Complexity: O(N) time to perform N/2 swaps
* Space complexity: O(N) to keep the recursion stack

```python
# recursive function 'helper' will receive and pass pointers as arguments

class Solution:
    def reverseString(self, s):
        def helper(left, right):
            if left < right: # Base case is if left >= right, do nothing
                s[left], s[right] = s[right], s[left] # otherwise, swap
                helper(left + 1, right - 1) #and call updated helper function

        helper(0, len(s) - 1) 
```

Java (Recursion, In-Place O(N) Space) *same complexities as above*

```python
class Solution{
    public void helper (char[] s, int left, int right) { #declares public helper method taking 3 parameters (char array 's', integers 'left' and 'right')
        if (left >= right) return; # base case
        char tmp= s[left]; #stores left index value in a temporary variable 'tmp'
        s[left++] = s[right]; #increments and swaps indexes
        s[right--] = tmp; #assigns value from 'tmp' to 'right' index, then decrements right
        helper(s, left, right); #recursive call swaps until base case met
    }

    public void reverseString(char[] s) { # declares public method that takes character array 's' as its parameter
        helper(s, 0, s.length - 1); # calls method to pass char array 's', initial 'left', 'right' indices
    }
}
```

# Sliding Window

## Subarrays

Given an array of positive integers nums and an integer k, return the number of subarrays where the product of all the elements in the subarray is strictly less than k.

For example, given the input nums = [10, 5, 2, 6], k = 100, the answer is 8. The subarrays with products less than k are:

[10], [5], [2], [6], [10, 5], [5, 2], [2, 6], [5, 2, 6]

```python
##Initialize class and variables with function
class Solution:
    def numSubarrayProductLessThanK(self, nums: list[int], k) -> int:
        if k <= 1: # constraint avoids infinite while loop since we only deal with positive integers
            return 0
        
        ans = left = 0 #pointer
        curr = 1 #product starts here, cannot be zero (multiplication)

        for right in range(len(nums)): #iterate over input
            curr *= nums[right] #add each element to current window by multiplying current window product by number
            while curr >= k: #condition broken
                curr //= nums[left] #remove from current window
                left += 1 #move pointer, goes infinitely if constraint is less than zero as no subarrays valid

            ans += right - left + 1 #total subarrays

        return ans
    
nums = [10, 5, 2, 6]
k = 100

Solution().numSubarrayProductLessThanK(nums, k)

#WORK ON EACH LOOP ITERATION AMORTIZED CONSTANT, runtime = O(n) where n = len(sums) and O(1) space
        
```

```python
def find_best_subarray(nums: list[int], k:int):
    
    #find sum of first window (1st k elements)
    curr = 0
    for i in range(k):
        curr += nums[i]

    #set answer to this value
    ans = curr
    #iterate over rest of windows
    for i in range(k, len(nums)):
        curr += nums[i] - nums[i - k]
        ans = max(ans, curr)

    return ans                                                                                                                                     

# The total for loop iterations is equal to n, where n = len(nums), and work done in each iteration is constant
# giving time complexity = O(n) using O(1) space
```
