```python
def find_length(nums, k):
    # curr is the current sum of the window
    left = curr = ans = 0 # initialize pointer variables: left pointer, current window sum, answer
    for right in range(len(nums)): # iterate over input
        curr += nums[right] # at each number, add current number to current window by incrementing current window sum by that value
        while curr > k: # as long as constraint broken (as long as sum of window is greater than the limit)
            curr -= nums[left] # remove from the left
            left += 1 # increment left (shrinks window)
        ans = max(ans, right - left + 1) # once window no longer breaks constraint, answer updates and yields length of window
        
    return ans # once entire input iterated answer returns
```

```python
nums = [3, 1, 2, 7, 4, 2, 1, 1, 5]
k = 8

find_length(nums, k)

# Time complexity is O(n) since all work done inside the for loop is amortized O(1), where *n* is the length of *nums*
# Space complexity is constant because we are only using 3 integer variables
```

### Example 2

Given a binary string *s*, you may choose up to one "0" and flip it to a "1". What is the length of the longest substring achievable that contains only "1"?

For example, given s = "1101100111", the answer is 5. If you perform the flip at index 2, the index becomes 1111100111.

Let s = "11001011"

```python
def find_length(s):
    # curr is the current number of zeros in the window
    left = curr = ans = 0 # initialize variables to zero : left pointer, current number of zeros in window, and the answer
    for right in range(len(s)): # iterate over input
        if s[right] == "0": # at each character, if it is zero
            curr += 1 # update window variable by adding one more zero to window
        
        while curr > 1: # as long as constraint is broken (more than one zero in window), shrink window
            if s[left] == "0": # if character at left pointer is zero
                curr -= 1 # update current variable by removing it
            left += 1 # shrink window by moving left forward
        # window no longer breaking constraint
        ans = max(ans, right - left + 1) # so answer updated with length of window
        
    return ans
```

```python
s = "11001011"

find_length(s)

# Problem runs in O(n) time, where n is the length of s, as the work done in each loop iteration is amortized constant
# Only a few integer variables are used as well, which means this algorithm uses O(1) space.
```

### Example 3

Find the number of subarrays that have a product less than *k*.

Given an array of positive integers nums and an integer k, return the number of subarrays where the product of all the elements in the subarray is strictly less than k.

For example, given the input nums = [10, 5, 2, 6], k = 100, the answer is 8. The subarrays with products less than k are:

[10], [5], [2], [6], [10, 5], [5, 2], [2, 6], [5, 2, 6]

```python
class Solution:
    def numSubarrayProductLessThanK(self, nums: list[int], k: int) -> int:
        
        if k <= 1: # BASE CASE REQUIRED, since condition will never be false when k <= 1, making left increase infinitely out of bounds
            return 0 # fixes infinite computation as no subarrays will fit this constraint
        
        ans = left = 0 # initialize variables as answer and left pointer starting at zero
        curr = 1 # current window product needs to start at 1 since multiplication cannot yield zero (based off conditions)
        
        for right in range(len(nums)): # iterate over the input
            curr *= nums[right] # at each element, add it to the current window by multiplying the current window product by the number
            while curr >= k: # while the condition is broken (current window product greater than or equal to k) need to remove from window
                curr //= nums[left] # with opposite of multiplication, division
                left += 1 # shrink window
                        
            ans += right - left + 1 # once window is valid, add number of subarrays that end at current index (right - left + 1)
            
        return ans # once whole input iterated, return answer
    
```

```python
nums = [10, 5 , 2, 6]
k = 100

Solution().numSubarrayProductLessThanK(nums,k)

# Work done in each loop iteration is amortized constant, so this algorithm has runtime of O(n), where n = length of nums, and O(1) space
```

### Example 4

Given an integer array *nums* and integer *k*, find the sum of the subarray with the largest sum whose length is *k*

```python
def find_best_subarray(nums: list[int], k:int):
    curr = 0
    for i in range(k): # find sum of first window
        curr += nums[i] # which is sum of first k elements
    #set answer to this value    
    ans = curr
    for i in range(k, len(nums)): #iterate over rest of windows
        curr += nums[i] - nums[i - k] #find each window sum by adding each element and removing previous element
        ans = max(ans, curr) #update answer at each iteration
        
    return ans
    
```

```python
k = 4
nums = [3, -1, 4, 12, -8, 5, 6]

find_best_subarray(nums, k)

# Total for loop iterations is equal to n, where n is the length of nums, and the work done in each iteration is constant, giving this algorithm a time complexity of O(n), using O(1) space.
# Must differentiate between examples and actual practice problem
```
